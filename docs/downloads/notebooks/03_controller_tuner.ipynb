Offline ranking generates a candidate, not permission to deploy it.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent))
from notebook_tools import simulate_line_controller
trials=[simulate_line_controller(kp,kd)['metrics'] for kp in (0.2,0.8,1.6,3.0) for kd in (0.0,0.1,0.4)]
ranked=sorted(trials,key=lambda m:(not m['passes'],m['rms_error_m'],m['saturation_fraction']))
ranked
```

```python
best=ranked[0]
second=simulate_line_controller(best['kp'],best['kd'],initial_error=-0.25)['metrics']
assert best['passes'] and second['passes']
best, second
```

```python
import matplotlib.pyplot as plt
for label,gains in {'best':(best['kp'],best['kd']),'sluggish':(0.2,0.0),'aggressive':(3.0,0.0)}.items():
    result=simulate_line_controller(*gains)
    plt.plot([r['time_s'] for r in result['rows']],[r['error_m'] for r in result['rows']],label=label)
plt.xlabel('time [s]'); plt.ylabel('error [m]'); plt.legend()
```
